```python
import imagehash
from PIL import Image
from collections import defaultdict

def calculate_image_hashes(data_dir):
    hash_dict = defaultdict(list)
    for root, dirs, files in os.walk(data_dir):
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                file_path = os.path.join(root, file)
                try:
                    img = Image.open(file_path)
                    hash_value = imagehash.phash(img)
                    hash_dict[hash_value].append(file_path)
                except Exception as e:
                    print(f"Error processing {file_path}: {e}")
    return hash_dict

def remove_duplicates(hash_dict):
    for hash_value, paths in hash_dict.items():
        if len(paths) > 1:
            print(f"Duplicate images found for hash {hash_value}:")
            for path in paths[1:]:
                print(f"Removing {path}")
                os.remove(path)

if __name__ == "__main__":
    hash_dict = calculate_image_hashes(data_dir_lf)
    remove_duplicates(hash_dict)
```

# Gabung semua data & EDA Simpel

```python
import os, glob, random

# Definisikan mapping class → index
classes = ['brown', 'normal', 'purple', 'scorch', 'shot hole']
class_to_idx = {c: i for i, c in enumerate(classes)}

# Path ke masing-masing folder
data_dirs = {
    'scorch': data_dir_lf,
    'normal': data_dir_nl,
    'brown': data_dir_bs,
    'purple': data_dir_pl,
    'shot hole': data_dir_sh
}

data_all = []

# Loop tiap kelas dan mengambil semua file
for c, path in data_dirs.items():
    for p in glob.glob(os.path.join(path, '*')):
        data_all.append((p, class_to_idx[c]))

# di Shuffle biar acak
random.Random(SEED_VALUE).shuffle(data_all)

print(f"Total items: {len(data_all)}")
print("Sample:", data_all[:5])
```

```python
import matplotlib.pyplot as plt
from PIL import Image

widths = []
heights = []
aspect_ratios = []

for c, path in data_dirs.items():   # loop tiap kelas dan path
    for root, dirs, files in os.walk(path):   # os.walk butuh string path
        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                file_path = os.path.join(root, file)
                try:
                    with Image.open(file_path) as img:
                        w, h = img.size
                        widths.append(w)
                        heights.append(h)
                        aspect_ratios.append(w/h)
                except Exception as e:
                    print(f"Error processing {file_path}: {e}")

print(f"Total images: {len(widths)}")

plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.scatter(widths, heights, alpha=0.5)
plt.title("Distribusi Resolusi Gambar")
plt.xlabel("Lebar")
plt.ylabel("Tinggi")

plt.subplot(1, 2, 2)
plt.hist(aspect_ratios, bins=20)
plt.title("Distribusi Aspect Ratio")
plt.xlabel("Aspect Ratio (Lebar/Tinggi)")
plt.ylabel("Frekuensi")
plt.show()
```

# Splitting Data

```python
batch_size = 32
img_height = 224
img_width = 224

random.Random(SEED_VALUE).shuffle(data_all)

# --- Split 70/15/15 ---
n_total = len(data_all)
n_train = int(0.7 * n_total)
n_val   = int(0.15 * n_total)

train_data = data_all[:n_train]
val_data   = data_all[n_train:n_train+n_val]
test_data  = data_all[n_train+n_val:]

print(f"Train: {len(train_data)}, Val: {len(val_data)}, Test: {len(test_data)}")

# --- Helper function untuk load image ---
def load_image(path, label):
    img = tf.io.read_file(path)
    img = tf.image.decode_jpeg(img, channels=3)
    img = tf.image.resize(img, [img_height, img_width])
    img = img / 255.0   # normalisasi
    return img, tf.one_hot(label, depth=len(classes))

# --- Convert list → tf.data.Dataset ---
# Separate paths and labels for each dataset
train_paths = [item[0] for item in train_data]
train_labels = [item[1] for item in train_data]
val_paths = [item[0] for item in val_data]
val_labels = [item[1] for item in val_data]
test_paths = [item[0] for item in test_data]
test_labels = [item[1] for item in test_data]

train_ds = tf.data.Dataset.from_tensor_slices((train_paths, train_labels)).map(load_image).batch(batch_size).shuffle(1000, seed=SEED_VALUE)
val_ds   = tf.data.Dataset.from_tensor_slices((val_paths, val_labels)).map(load_image).batch(batch_size)
test_ds  = tf.data.Dataset.from_tensor_slices((test_paths, test_labels)).map(load_image).batch(batch_size)
```

# Augmentasi data

```python
from tensorflow.keras import layers
import tensorflow as tf

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.2), # Rotasi acak hingga 20%
    layers.RandomZoom(0.2),     # Zoom in/out acak hingga 20%
    layers.RandomContrast(0.1)  # Mengubah kontras acak untuk simulasi pencahayaan
], name="data_augmentation_layer")
```

```python
# Menerapkan augmentasi
aug_train_ds = train_ds.map(
    lambda x, y: (data_augmentation(x, training=True), y),
    num_parallel_calls=tf.data.AUTOTUNE
)

# Optimasi performa pipeline
aug_train_ds = aug_train_ds.prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=tf.data.AUTOTUNE)
test_ds = test_ds.prefetch(buffer_size=tf.data.AUTOTUNE)
```

# Pembuatan Baseline Model

Baseline model dibangun menggunakan arsitektur AlexNet murni secara manual. Berdasarkan diagram, input awal adalah 224x224x3. Model menggunakan Convolutional layer dengan ukuran filter dan stride yang spesifik (misal layer pertama: kernel 11x11, stride 4, 96 filter). Semua Convolutional layer menggunakan aktivasi ReLU, diikuti oleh Max Pooling 3x3 dengan stride 2. Bagian Dense network menggunakan dua layer berukuran 4096 neuron dengan Dropout 0.5. Layer output, yang aslinya berisi 1000 neuron, diadaptasi menjadi 5 neuron dengan aktivasi softmax karena dataset daun ceri ini memiliki 5 kelas penyakit.

```python
from tensorflow.keras import layers, models

def build_alexnet(num_classes=5):
    model = models.Sequential()

    model.add(layers.Conv2D(96, (11, 11), strides=4, activation='relu', input_shape=(224, 224, 3)))
    model.add(layers.MaxPooling2D(pool_size=(3, 3), strides=2))

    model.add(layers.Conv2D(256, (5, 5), padding='same', activation='relu'))
    model.add(layers.MaxPooling2D(pool_size=(3, 3), strides=2))

    model.add(layers.Conv2D(384, (3, 3), padding='same', activation='relu'))
    model.add(layers.Conv2D(384, (3, 3), padding='same', activation='relu'))
    model.add(layers.Conv2D(256, (3, 3), padding='same', activation='relu'))

    model.add(layers.MaxPooling2D(pool_size=(3, 3), strides=2))

    model.add(layers.Flatten())

    model.add(layers.Dense(4096, activation='relu'))
    model.add(layers.Dropout(0.5))

    model.add(layers.Dense(4096, activation='relu'))
    model.add(layers.Dropout(0.5))

    model.add(layers.Dense(num_classes, activation='softmax'))

    return model

baseline_model = build_alexnet(num_classes=5)
baseline_model.compile(optimizer='adam',
                       loss='categorical_crossentropy',
                       metrics=['accuracy'])

baseline_model.summary()
```

```python
# TRAINING MODEL

early_stopping_base = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True,
    verbose=1
)

baseline_history = baseline_model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=[early_stopping_base]
)
```

# C. Modifikasi Model

```python
from sklearn.utils.class_weight import compute_class_weight
from tensorflow.keras import layers, models
import numpy as np
import tensorflow as tf

# hitung class weights
print("Menghitung class weights...")
train_labels = []
for images, labels in train_ds: # Gunakan train_ds (sebelum augmentasi batch) untuk menghitung label
    train_labels.extend(np.argmax(labels.numpy(), axis=1))

# Menghitung bobot secara otomatis agar kelas minoritas mendapat bobot lebih besar
classes = np.unique(train_labels)
weights = compute_class_weight(class_weight='balanced', classes=classes, y=train_labels)
class_weight_dict = dict(zip(classes, weights))
print("Class Weights yang digunakan:", class_weight_dict)


# MEMBUAT CUSTOM CNN MODEL
def build_custom_cnn(num_classes=5):
    model = models.Sequential(name="Custom_CNN_Modified")

    # Block 1
    model.add(layers.Conv2D(64, (3, 3), padding='same', activation='relu', input_shape=(224, 224, 3)))
    model.add(layers.BatchNormalization())
    model.add(layers.Conv2D(64, (3, 3), padding='same', activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D((2, 2)))
    model.add(layers.Dropout(0.25))

    # Block 2
    model.add(layers.Conv2D(128, (3, 3), padding='same', activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D((2, 2)))
    model.add(layers.Dropout(0.25))

    # Block 3
    model.add(layers.Conv2D(256, (3, 3), padding='same', activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.MaxPooling2D((2, 2)))
    model.add(layers.Dropout(0.25))

    # Fully Connected Layer
    model.add(layers.Flatten())
    model.add(layers.Dense(512, activation='relu'))
    model.add(layers.BatchNormalization())
    model.add(layers.Dropout(0.5))

    # Output Layer
    model.add(layers.Dense(num_classes, activation='softmax'))

    return model

modified_model = build_custom_cnn(num_classes=5)
modified_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
                       loss='categorical_crossentropy',
                       metrics=['accuracy'])

modified_model.summary()


# 3. MENGATUR EARLY STOPPING
# Berhenti jika val_loss tidak membaik selama 7 epoch berturut-turut
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor='val_loss',
    patience=7,
    restore_best_weights=True,
    verbose=1
)
```

keunggulan modified model:
1. BatchNormalization --> membuat training lebih stabil, loss lebih cepat turun
2. Dropout bertahap --> (harapan) kontrol overfitting lebih baik
3. Filter kecil (3x3) --> fokus menangkap detail lokal

Fokus beberapa metrik ini untuk mengantisipasi kesalahan memprediksi daun sakit menjadi daun sehat bisa berakibat pada penyebaran kesimpulan model mana yang memiliki generalisasi yang lebih baik.

```python
from sklearn.metrics import classification_report
import numpy as np

y_true = []
y_pred_baseline = []
y_pred_modified = []

for images, labels in test_ds:
    y_true.extend(np.argmax(labels.numpy(), axis=1))

    baseline_preds = baseline_model.predict(images, verbose=0)
    y_pred_baseline.extend(np.argmax(baseline_preds, axis=1))

    modified_preds = modified_model.predict(images, verbose=0)
    y_pred_modified.extend(np.argmax(modified_preds, axis=1))

print("=== Evaluasi Baseline Model (AlexNet) ===")
print(classification_report(y_true, y_pred_baseline, target_names=["Brown Spot", "Leaf Scorch", "Normal", "Purple Spot", "Shot Hole"]))

print("\n=== Evaluasi Modified Model (EfficientNet) ===")
print(classification_report(y_true, y_pred_modified, target_names=["Brown Spot", "Leaf Scorch", "Normal", "Purple Spot", "Shot Hole"]))
```

2. butuh tuning lebih hati-hati
